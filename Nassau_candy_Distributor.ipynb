```python
# Calculate missing date components first
df["Order_Month"] = df["Order Date"].dt.month
df["Order_Year"] = df["Order Date"].dt.year

# Create the ML dataset
ml_df = df[
    [
        "Product Name",
        "Factory",
        "Region",
        "Ship Mode",
        "Order_Month",
        "Order_Year",
        "Lead_Time_Days"
    ]
].copy()

print("ML dataset shape:", ml_df.shape)
print("\nMissing values:")
print(ml_df.isna().sum())
print("\nSample:")
print(ml_df.head())
```

```python
from sklearn.model_selection import train_test_split

# Features
X = ml_df[
    [
        "Product Name",
        "Factory",
        "Region",
        "Ship Mode",
        "Order_Month",
        "Order_Year"
    ]
]

# Target
y = ml_df["Lead_Time_Days"]

# Split the data
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

print("Training rows:", X_train.shape[0])
print("Testing rows:", X_test.shape[0])

print("\nTraining features:", X_train.shape[1])
print("Testing features:", X_test.shape[1])
```

```python
from sklearn.preprocessing import OneHotEncoder

categorical_features = [
    "Product Name",
    "Factory",
    "Region",
    "Ship Mode"
]

# Create encoder
encoder = OneHotEncoder(
    handle_unknown="ignore",
    sparse_output=False
)

# Fit ONLY on training data
X_train_encoded = encoder.fit_transform(
    X_train[categorical_features]
)

# Transform test data using the same encoder
X_test_encoded = encoder.transform(
    X_test[categorical_features]
)

# Get encoded column names
encoded_columns = encoder.get_feature_names_out(
    categorical_features
)

print("Encoded training shape:", X_train_encoded.shape)
print("Encoded testing shape:", X_test_encoded.shape)

print("\nEncoded features:")
print(list(encoded_columns))
```

```python
import pandas as pd

# Convert encoded arrays into DataFrames
X_train_encoded_df = pd.DataFrame(
    X_train_encoded,
    columns=encoded_columns,
    index=X_train.index
)

X_test_encoded_df = pd.DataFrame(
    X_test_encoded,
    columns=encoded_columns,
    index=X_test.index
)

# Add numerical features
X_train_final = pd.concat(
    [
        X_train_encoded_df,
        X_train[["Order_Month", "Order_Year"]]
    ],
    axis=1
)

X_test_final = pd.concat(
    [
        X_test_encoded_df,
        X_test[["Order_Month", "Order_Year"]]
    ],
    axis=1
)

print("Final training shape:", X_train_final.shape)
print("Final testing shape:", X_test_final.shape)

print("\nTotal features:", X_train_final.shape[1])
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score
import numpy as np
import pandas as pd

# Ensure ml_df and other preparation variables are defined
if 'ml_df' not in globals():
    # Reload raw dataset
    df = pd.read_csv("/content/Nassau Candy Distributor.csv")

    # Correctly parse dates as DD-MM-YYYY
    df["Order Date"] = pd.to_datetime(df["Order Date"], format="%d-%m-%Y")
    df["Ship Date"] = pd.to_datetime(df["Ship Date"], format="%d-%m-%Y")

    # Calculate Lead Time
    df["Lead_Time_Days"] = (df["Ship Date"] - df["Order Date"]).dt.days

    # Map Products to Factories
    factory_mapping = {
        "Wonka Bar - Nutty Crunch Surprise": "Lot's O' Nuts",
        "Wonka Bar - Fudge Mallows": "Lot's O' Nuts",
        "Wonka Bar -Scrumdiddlyumptious": "Lot's O' Nuts",
        "Wonka Bar - Milk Chocolate": "Wicked Choccy's",
        "Wonka Bar - Triple Dazzle Caramel": "Wicked Choccy's",
        "Laffy Taffy": "Sugar Shack",
        "SweeTARTS": "Sugar Shack",
        "Nerds": "Sugar Shack",
        "Fun Dip": "Sugar Shack",
        "Fizzy Lifting Drinks": "Sugar Shack",
        "Everlasting Gobstopper": "Secret Factory",
        "Hair Toffee": "The Other Factory",
        "Lickable Wallpaper": "Secret Factory",
        "Wonka Gum": "Secret Factory",
        "Kazookles": "The Other Factory"
    }
    df["Factory"] = df["Product Name"].map(factory_mapping)
    df["Order_Month"] = df["Order Date"].dt.month
    df["Order_Year"] = df["Order Date"].dt.year

    # Create ML dataframe
    ml_df = df[["Product Name", "Factory", "Region", "Ship Mode", "Order_Month", "Order_Year", "Lead_Time_Days"]].copy()

if 'X_train' not in globals() or 'X_test' not in globals():
    from sklearn.model_selection import train_test_split
    X = ml_df[["Product Name", "Factory", "Region", "Ship Mode", "Order_Month", "Order_Year"]]
    y = ml_df["Lead_Time_Days"]
    X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.20, random_state=42)

if 'X_train_final' not in globals() or 'X_test_final' not in globals():
    from sklearn.preprocessing import OneHotEncoder
    categorical_features = ["Product Name", "Factory", "Region", "Ship Mode"]
    encoder = OneHotEncoder(handle_unknown="ignore", sparse_output=False)

    X_train_encoded = encoder.fit_transform(X_train[categorical_features])
    X_test_encoded = encoder.transform(X_test[categorical_features])
    encoded_columns = encoder.get_feature_names_out(categorical_features)

    X_train_encoded_df = pd.DataFrame(X_train_encoded, columns=encoded_columns, index=X_train.index)
    X_test_encoded_df = pd.DataFrame(X_test_encoded, columns=encoded_columns, index=X_test.index)

    X_train_final = pd.concat([X_train_encoded_df, X_train[["Order_Month", "Order_Year"]]], axis=1)
    X_test_final = pd.concat([X_test_encoded_df, X_test[["Order_Month", "Order_Year"]]], axis=1)

# Create Linear Regression model
lr_model = LinearRegression()

# Train the model
lr_model.fit(X_train_final, y_train)

# Predict on test data
lr_predictions = lr_model.predict(X_test_final)

# Evaluate the model
lr_rmse = np.sqrt(mean_squared_error(y_test, lr_predictions))
lr_mae = mean_absolute_error(y_test, lr_predictions)
lr_r2 = r2_score(y_test, lr_predictions)

print("Linear Regression Results")
print("-------------------------")
print("RMSE:", round(lr_rmse, 2))
print("MAE :", round(lr_mae, 2))
print("R²  :", round(lr_r2, 4))
```

```python
from sklearn.ensemble import RandomForestRegressor

# Create Random Forest model
rf_model = RandomForestRegressor(
    n_estimators=200,
    random_state=42,
    n_jobs=-1
)

# Train the model
rf_model.fit(X_train_final, y_train)

# Predict
rf_predictions = rf_model.predict(X_test_final)

# Evaluate
rf_rmse = np.sqrt(
    mean_squared_error(y_test, rf_predictions)
)

rf_mae = mean_absolute_error(
    y_test,
    rf_predictions
)

rf_r2 = r2_score(
    y_test,
    rf_predictions
)

print("Random Forest Results")
print("---------------------")
print("RMSE:", round(rf_rmse, 2))
print("MAE :", round(rf_mae, 2))
print("R²  :", round(rf_r2, 4))
```

```python
from sklearn.ensemble import GradientBoostingRegressor

# Create Gradient Boosting model
gb_model = GradientBoostingRegressor(
    n_estimators=200,
    learning_rate=0.05,
    max_depth=3,
    random_state=42
)

# Train
gb_model.fit(X_train_final, y_train)

# Predict
gb_predictions = gb_model.predict(X_test_final)

# Evaluate
gb_rmse = np.sqrt(
    mean_squared_error(y_test, gb_predictions)
)

gb_mae = mean_absolute_error(
    y_test,
    gb_predictions
)

gb_r2 = r2_score(
    y_test,
    gb_predictions
)

print("Gradient Boosting Results")
print("-------------------------")
print("RMSE:", round(gb_rmse, 2))
print("MAE :", round(gb_mae, 2))
print("R²  :", round(gb_r2, 4))
```

```python
import pandas as pd

# Fallback: Reconstruct recommendation_table if it's missing from the active session
if 'recommendation_table' not in globals():
    print("Variable 'recommendation_table' not found in memory. Reconstructing...")
    if 'df' not in globals():
        df = pd.read_csv("/content/Nassau Candy Distributor.csv")
        df["Order Date"] = pd.to_datetime(df["Order Date"], format="%d-%m-%Y")
        df["Ship Date"] = pd.to_datetime(df["Ship Date"], format="%d-%m-%Y")
        df["Lead_Time_Days"] = (df["Ship Date"] - df["Order Date"]).dt.days
        factory_mapping = {
            "Wonka Bar - Nutty Crunch Surprise": "Lot's O' Nuts",
            "Wonka Bar - Fudge Mallows": "Lot's O' Nuts",
            "Wonka Bar -Scrumdiddlyumptious": "Lot's O' Nuts",
            "Wonka Bar - Milk Chocolate": "Wicked Choccy's",
            "Wonka Bar - Triple Dazzle Caramel": "Wicked Choccy's",
            "Laffy Taffy": "Sugar Shack",
            "SweeTARTS": "Sugar Shack",
            "Nerds": "Sugar Shack",
            "Fun Dip": "Sugar Shack",
            "Fizzy Lifting Drinks": "Sugar Shack",
            "Everlasting Gobstopper": "Secret Factory",
            "Hair Toffee": "The Other Factory",
            "Lickable Wallpaper": "Secret Factory",
            "Wonka Gum": "Secret Factory",
            "Kazookles": "The Other Factory"
        }
        df["Factory"] = df["Product Name"].map(factory_mapping)
        df["Order_Month"] = df["Order Date"].dt.month
        df["Order_Year"] = df["Order Date"].dt.year

    if 'best_factory_recommendations' not in globals():
        from sklearn.preprocessing import OneHotEncoder
        from sklearn.model_selection import train_test_split
        from sklearn.ensemble import GradientBoostingRegressor

        ml_df = df[["Product Name", "Factory", "Region", "Ship Mode", "Order_Month", "Order_Year", "Lead_Time_Days"]].copy()
        X = ml_df[["Product Name", "Factory", "Region", "Ship Mode", "Order_Month", "Order_Year"]]
        y = ml_df["Lead_Time_Days"]
        X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.20, random_state=42)

        categorical_features = ["Product Name", "Factory", "Region", "Ship Mode"]
        encoder = OneHotEncoder(handle_unknown="ignore", sparse_output=False)
        encoder.fit(X_train[categorical_features])
        encoded_columns = encoder.get_feature_names_out(categorical_features)

        scenario_base = df[["Product Name", "Region", "Ship Mode", "Order_Month", "Order_Year", "Factory"]].copy()
        scenario_base = scenario_base.rename(columns={"Factory": "Current Factory"})
        all_factories = df["Factory"].unique().tolist()
        scenario_orders = scenario_base.merge(pd.DataFrame({"Scenario Factory": all_factories}), how="cross")

        scenario_orders["Scenario Type"] = scenario_orders.apply(
            lambda row: "Current" if row["Scenario Factory"] == row["Current Factory"] else "Alternative", axis=1
        )

        scenario_cat = scenario_orders[["Product Name", "Scenario Factory", "Region", "Ship Mode"]].copy()
        scenario_cat = scenario_cat.rename(columns={"Scenario Factory": "Factory"})
        scenario_encoded = encoder.transform(scenario_cat)
        scenario_encoded_df = pd.DataFrame(scenario_encoded, columns=encoded_columns, index=scenario_orders.index)
        scenario_final = pd.concat([scenario_encoded_df, scenario_orders[["Order_Month", "Order_Year"]]], axis=1)

        X_train_encoded = encoder.transform(X_train[categorical_features])
        X_train_encoded_df = pd.DataFrame(X_train_encoded, columns=encoded_columns, index=X_train.index)
        X_train_final = pd.concat([X_train_encoded_df, X_train[["Order_Month", "Order_Year"]]], axis=1)

        gb_model = GradientBoostingRegressor(n_estimators=200, learning_rate=0.05, max_depth=3, random_state=42)
        gb_model.fit(X_train_final, y_train)

        scenario_orders["Predicted_Lead_Time"] = gb_model.predict(scenario_final)
        scenario_orders["Predicted_Improvement_Days"] = (
            scenario_orders.groupby("Product Name")["Predicted_Lead_Time"].transform("first")
            - scenario_orders["Predicted_Lead_Time"]
        )

        best_factory_recommendations = (
            scenario_orders
            .loc[scenario_orders.groupby("Product Name")["Predicted_Lead_Time"].idxmin()]
            [["Product Name", "Current Factory", "Scenario Factory", "Predicted_Lead_Time", "Predicted_Improvement_Days"]]
            .reset_index(drop=True)
        )
        best_factory_recommendations["Improvement_%"] = (
            best_factory_recommendations["Predicted_Improvement_Days"]
            / (best_factory_recommendations["Predicted_Lead_Time"] + best_factory_recommendations["Predicted_Improvement_Days"])
        ) * 100

    recommendation_table = best_factory_recommendations.sort_values(
        by="Improvement_%",
        ascending=False
    ).reset_index(drop=True)

# Calculate reallocation opportunities where the recommended factory differs from the current factory
actual_reallocations = (
    recommendation_table["Current Factory"] != recommendation_table["Scenario Factory"]
).sum()

print("Actual Reallocation Opportunities:", actual_reallocations)
```

```python
reallocation_risk = recommendation_table[
    recommendation_table["Current Factory"] != recommendation_table["Scenario Factory"]
].copy()

reallocation_risk["Risk_Level"] = pd.cut(
    reallocation_risk["Improvement_%"],
    bins=[-float("inf"), 3, 7, float("inf")],
    labels=["High Risk", "Medium Risk", "Low Risk"]
)

print(
    reallocation_risk[
        [
            "Product Name",
            "Current Factory",
            "Scenario Factory",
            "Improvement_%",
            "Risk_Level"
        ]
    ].round(2)
)
```
